## Driver vs executor memory

```python
for item in ['spark.driver.memory', 'spark.executor.memory', 'spark.memory.fraction']:
    print(f'{item}: {spark.conf.get(item, "default")}')
```

## GC under load

```python
trips = spark.read.parquet(*config.yellow_taxi_paths())
heavy = trips.groupBy('PULocationID', 'DOLocationID', 'VendorID').count()
_, t = timed_action(heavy)
print(f'Aggregation: {t:.2f}s — open Executors tab and inspect GC Time')
```

## Key Takeaways

1. Read GC Time on the Executors tab before changing JVM flags.
2. Driver OOM and executor OOM show up in different UI places and logs.
3. `spark.memory.fraction` splits on-heap space between execution and storage.
